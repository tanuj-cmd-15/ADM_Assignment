```python
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.model_selection import train_test_split

data = {
    
    'Loan Token': ['Y', 'Y', 'Y', 'N', 'N', 'N', 'N', 'Y', 'N'],
    'Taxable Income': ['Y', 'Y', 'N', 'N', 'Y', 'Y', 'N', 'N', 'Y'],
    'Target Class': ['Positive', 'Positive', 'Negative', 'Positive', 'Negative', 'Negative', 'Negative', 'Positive', 'Negative']
}

df = pd.DataFrame(data)

df_encoded = df.copy()
df_encoded['Loan Token'] = df['Loan Token'].map({'Y': 1, 'N': 0})
df_encoded['Taxable Income'] = df['Taxable Income'].map({'Y': 1, 'N': 0})
df_encoded['Target Class'] = df['Target Class'].map({'Positive': 1, 'Negative': 0})

X = df_encoded[['Loan Token', 'Taxable Income']]
y = df_encoded['Target Class']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Calculate entropy
def entropy(probabilities):
    return -sum(p * np.log2(p) for p in probabilities if p > 0)

# Calculate information gain
def info_gain(dataset, feature, target):
    total_entropy = entropy(dataset[target].value_counts(normalize=True))
    values = dataset[feature].unique()
    feature_entropy = 0
    for value in values:
        subset = dataset[dataset[feature] == value]
        prob = len(subset) / len(dataset)
        feature_entropy += prob * entropy(subset[target].value_counts(normalize=True))
    return total_entropy - feature_entropy

# Entropy and Information Gain Calculation
features = ['Loan Token', 'Taxable Income']
print("Entropy and Information Gain:")
for feature in features:
    gain = info_gain(df_encoded, feature, 'Target Class')
    print(f"Feature: {feature}, Information Gain: {gain:.4f}")


clf = DecisionTreeClassifier(random_state=42)
clf.fit(X_train, y_train)

# Evaluate performance
preds = clf.predict(X_test)
print("\nDecision Tree Performance (No Pruning):")
print(f"Accuracy: {accuracy_score(y_test, preds):.2f}")
print(f"Precision: {precision_score(y_test, preds):.2f}")
print(f"Recall: {recall_score(y_test, preds):.2f}")
print(f"F1-Score: {f1_score(y_test, preds):.2f}")

pruned_clf = DecisionTreeClassifier(max_depth=2, random_state=42)
pruned_clf.fit(X_train, y_train)

pruned_preds = pruned_clf.predict(X_test)
print("\nDecision Tree Performance (Pruned):")
print(f"Accuracy: {accuracy_score(y_test, pruned_preds):.2f}")
print(f"Precision: {precision_score(y_test, pruned_preds):.2f}")
print(f"Recall: {recall_score(y_test, pruned_preds):.2f}")
print(f"F1-Score: {f1_score(y_test, pruned_preds):.2f}")


```
